Which age group purchases the most Clothing, Electronics, and Beauty products top10?
### Objective

To identify the purchasing patterns of different age groups across product categories of top 10.

```python
df_product_by_age=df.groupby(['Age','Product Category']).size().sort_values(ascending=False).reset_index(name="Count").head(10)
df_product_by_age
```

```python
#visualizing the data
fig=px.bar(data_frame=df_product_by_age,x="Age",y="Count",color="Product Category",barmode="group",title="Product Purchases by Age Group",text_auto=True,color_discrete_sequence=px.colors.qualitative.Set2,template="plotly_white")
fig.update_layout(
    width=1400,
    height=600,
    bargap=1,
    title="Top 10 Customer Age-product Category Combinations by Purchase Frequency",
    title_x=0.5
)
fig.update_traces(
    width=1
)
fig.update_xaxes(
    title="Age(Years)",
    dtick=4
)
fig.update_yaxes(
    title="Purchase Count"
)
fig.update_layout(
    title_font=dict(
        family="Arial",
        size=24,
        color="#1F77B4"
    ),
    xaxis_title_font=dict(
        family="Arial",
        size=18,
        color="#2F4F4F"
    ),
    yaxis_title_font=dict(
        family="Arial",
        size=18,
        color="#2F4F4F"
    )
)
fig.show()
```

📌 Insights
Electronics appears most frequently among the top 10 age–product category combinations, indicating strong purchase activity for this category across multiple age groups.

Customers aged 51 years recorded the highest purchase count (14) for Electronics, making it the most frequent age–category combination in the dataset.

Clothing also shows high demand, with customers aged 43 and 64 each making 13 purchases.

Beauty products reached a maximum purchase count of 13 among customers aged 46, suggesting strong interest in this age group.

Most of the top-performing combinations are concentrated between the ages of 30 and 64, indicating that middle-aged customers contribute significantly to product purchases.

The purchase counts among the top 10 combinations range from 12 to 14, showing relatively consistent demand across these leading age–category combinations.

Which product category contributes the highest share of total units sold?

🎯 Objective
To analyze the distribution of total units sold across different product categories and identify which category contributes the largest share of overall sales. This analysis helps understand customer purchasing preferences and supports inventory planning and product management decisions.


```python
# Group By : Product Category
# Aggregate : Quantity
# Function : Sum
df_Quantity=df.groupby(['Product Category'])['Quantity'].sum().reset_index()
df_Quantity
```

```python
fig=px.pie(data_frame=df_Quantity,names="Product Category",values="Quantity",hole=0.6,hover_name='Product Category',color_discrete_sequence=px.colors.qualitative.Set2,template="plotly_white")
fig.update_traces(
    textinfo="percent+label"
)
fig.update_layout(
    legend=dict(
        y=-0.2,
        x=0.5,
        xanchor="center"
    ),
    title="Distribution of Total Units Sold by Product Category",
     title_font=dict(
            family="Arial",
            size=22,
            color="#1F77B4"
        ),
        title_x=0.5,
        showlegend=False,
        height=600,
        width=1000
)
fig.show()
```

📌Insights

Clothing contributes the largest share of total units sold (35.6%), making it the highest-selling product category in the dataset.

Electronics accounts for 33.8% of total units sold, closely following Clothing, indicating strong and consistent customer demand.

Beauty represents 30.7% of total units sold, making it the least sold product category, although the difference compared to the other categories is relatively small.

The sales distribution is fairly balanced across all three product categories, with each contributing approximately one-third of the total units sold, suggesting diversified customer purchasing behavior.

The difference between the highest-selling category (Clothing) and the lowest-selling category (Beauty) is only 4.9 percentage points, indicating that customer demand is relatively evenly distributed among all product categories.

Since all three product categories contribute significantly to total sales, maintaining balanced inventory levels across Clothing, Electronics, and Beauty would help meet customer demand and reduce the risk of stock shortages.

How do purchasing patterns differ between male and female customers across different product categories?

🎯 Objective

To analyze how product purchases vary between male and female customers across different product categories. This analysis helps identify customer preferences and supports targeted marketing and inventory planning.

```python
df_purchase=df.groupby(['Gender','Product Category'])['Quantity'].sum().reset_index()
df_purchase
```

```python
fig=px.bar(data_frame=df_purchase,x='Product Category',y="Quantity",barmode="group",color="Gender",    color_discrete_map={
      "Male":"#4C78A8",
    "Female":"#F58518"
})
fig.update_layout(
    title="Product Purchases by Gender Across Product Categories",
    title_font=dict(
            size=22,
            color="#1F77B4",
            family="Arial"
        ),
    showlegend=False,
    width=1000,
    height=600,
    title_x=0.5

)
fig.update_yaxes(
    title="Total Units Purchased"
)
fig.show()
```

📌 Key Insights

Female customers purchased more Beauty products than male customers, indicating a stronger preference for Beauty products among female shoppers.

Male customers purchased slightly more Clothing products than female customers, making Clothing the most purchased category among male customers.

Female customers recorded higher purchases in Electronics compared to male customers, suggesting strong customer interest in this category.

Clothing and Electronics were the two most purchased product categories, with both genders contributing significantly to their total sales.

The difference in purchase quantity between male and female customers is relatively small across most product categories, indicating balanced purchasing behavior between genders.

The purchasing preferences vary across product categories, highlighting opportunities for businesses to design targeted marketing campaigns and promotional offers for different customer segments.

Does customer age influence clothing purchase quantity?

🎯 Objective

To examine whether there is a relationship between customer age and the quantity of clothing products purchased. This analysis helps determine whether age is a significant factor influencing clothing purchasing behavior.

```python
df_cutomer=df[df['Product Category']=="Clothing"]
df_cutomer
```

```python
df_cutomer.groupby('Age')['Quantity'].sum().reset_index()
```

```python
fig=px.scatter(data_frame=df_cutomer,x='Age',y="Quantity",trendline='ols',title="Relationship Between Age and Clothing Purchases")
fig.update_traces(visible=False,selector=dict(mode="markers"))
fig.update_layout(
    
    width=600
)
fig.show()
```

📌 Key Insights
The regression trendline shows a slight negative relationship between customer age and clothing purchase quantity.

As customer age increases, the average clothing purchase quantity tends to decrease slightly, although the decline is minimal.

The gentle slope of the regression line indicates that age has only a weak influence on clothing purchase quantity.

No strong upward or downward trend is observed, suggesting that customers across different age groups purchase clothing in relatively similar quantities.

Based on this analysis, customer age alone does not appear to be a major factor in determining clothing purchase quantity, and additional factors such as customer preferences, pricing, or promotions may have a greater impact on purchasing behavior.

During which month were the highest sales recorded?

🎯 Objective

To analyze monthly sales performance by identifying the total sales generated in each month. This analysis helps uncover seasonal sales trends, determine the best-performing month, and support better business planning, inventory management, and marketing strategies.


```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

df_month["Month"] = pd.Categorical(
    df_month["Month"],
    categories=month_order,
    ordered=True
)

df_month = df_month.sort_values("Month")
```

```python
df_month=df.groupby('Month')['Total Amount'].sum().reset_index()
fig=px.line(data_frame=df_month,x='Month',y="Total Amount")
fig.show()
```
